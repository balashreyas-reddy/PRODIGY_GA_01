```python
from transformers import pipeline

# Load GPT-2 model
generator = pipeline(
    "text-generation",
    model="gpt2"
)

# Take prompt from user
prompt = input("Enter your prompt: ")

# Generate text
result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1
)

# Display output
print("\nGenerated Text:\n")
print(result[0]["generated_text"])
```
